```python
from pyspark.sql.functions import col, when

df = spark.table(
    "retail_catalog.sales_demo.raw_sales"
)

# Filter invalid records
clean_df = df.filter(
    col("sales_amount") > 0
)

# Add calculated field
transformed_df = clean_df.withColumn(
    "sales_category",
    when(col("sales_amount") >= 20000, "High Value")
    .otherwise("Standard")
)

# Aggregation
agg_df = transformed_df.groupBy("region").sum("sales_amount")

display(agg_df)

transformed_df.write.mode("overwrite").saveAsTable(
    "retail_catalog.sales_demo.transformed_sales"
)
```
