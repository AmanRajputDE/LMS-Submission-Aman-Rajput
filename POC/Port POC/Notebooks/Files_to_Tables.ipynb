```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.read.csv(
    "abfss://POC@onelake.dfs.fabric.microsoft.com/bronze_lh.Lakehouse/Files/db_date/fact_gate_transaction",header=True
)
```

```python
display(df)
```

```python
df = df.withColumn(
    "EntryTime",
    to_timestamp(col("EntryTime"), "yyyy-MM-dd HH:mm:ss")
).withColumn(
    "ExitTime",
    to_timestamp(col("ExitTime"), "yyyy-MM-dd HH:mm:ss")
).withColumn(
    "ProcessingMinutes",
    col("ProcessingMinutes").cast("int")
)
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("fact_gate_transaction")
```
